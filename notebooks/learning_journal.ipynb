## 1 · How long have I been curious?

```python
from datetime import date

started = date(2021, 9, 1)   # roughly when it all began
days = (date.today() - started).days
print(f"{days:,} days of curiosity")
```

## 2 · One honest week
The same numbers as the website's mini chart. Update both together.

```python
import matplotlib.pyplot as plt

week = {"Python": 7, "FastAPI": 5, "Docs": 4, "Drawing": 3, "ML": 2}
colors = ["#FFC4D6", "#F0709B", "#DCCDFF", "#8B5CC9", "#C3F0DD"]

fig, ax = plt.subplots(figsize=(7, 3.2))
ax.barh(list(week), list(week.values()), color=colors)
ax.invert_yaxis()
ax.set_xlabel("hours")
ax.spines[["top", "right"]].set_visible(False)
ax.set_title("Learning hours this week", loc="left")
plt.tight_layout()
plt.show()
```

## 3 · Talk to my own site
Run the server first (`uvicorn app.main:app --reload`), then:

```python
import json
from urllib.request import urlopen

for cell in ["days", "week", "explore"]:
    with urlopen(f"http://127.0.0.1:8000/api/notebook/{cell}?lang=ar") as r:
        print(cell, "→", json.loads(r.read()))
```
